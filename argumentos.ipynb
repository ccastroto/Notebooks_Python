## 1. Arguments and Shared References

### Paso de objetos inmutables:

![metafora pulpo variables](images/00%20metafora%20pulpo%20variables.png)

![referencia objetos Python](images/01%20Variables,%20objetos%20y%20referencias.png)

![objeto con dos referencias](images/02%20objeto%20con%20dos%20referencias.png)

```python
def f(a):	    	# a is assigned to (references) the passed object
	a = 99 			# Changes local variable a only

b = 88		
f(b)				# a and b both reference same 88 initially
print("b = ", b)	# b is not changed
```

### Paso de objetos mutables:


```python

def changer(a, b):	    # Arguments assigned references to objects
		a = 2			# Changes local name's value only
		b[0] = 'spam' 	# Changes shared object in place	
					
x = 1
l = [1, 2]
changer(x, l)           # Caller: Pass immutable and mutable objects
print("l = ", l)        # ['spam', 2]  # x is unchanged, l is different!
assert x == 1, "x es 1"
```

### Cómo evitar que una rutina modifique los elementos mutables:

#### - Pasar una copia

```python

l = [1, 2]
changer(x, l[:]) 	# Pass a copy, so our 'l' does not change 

print("l = ", l) 
```

#### - Utilizar invariantes (en las estructuras de datos de entrada)

```python
l = [1, 2]
def changer(a, b):
    b = b[:]		# Copy input list so we don't impact caller
    a = 2
    b[0] = 'spam'	# Changes our list copy only

changer(x, l)
assert l == [1, 2], "l es una invariante"
print("l = ", l) 
```

#### - Convertir el objeto mutable a un objeto inmutable:


```python

l = [1, 2]
def changer(a, b):	   
		a = 2			
		b[0] = 'spam'   # TypeError: 'tuple' object does not support item assignment

changer(x, tuple(l))	# Pass a tuple, so changes are errors

print("l = ", l)
assert l == [1, 2], "l es una invariante"
```

## 2 Special Argument-Matching Modes

#### ver figura 06 en Drive

Ten cuidado de no confundir la sintaxis especial `nombre=valor` en la cabecera de una función y en una llamada a función; en la llamada significa que se está pasando un argumento por nombre (keyword argument), mientras que en la cabecera indica un valor por defecto para un argumento opcional.

### Keyword and Default Examples

#### 1. Comportamiento por defecto:

```python
def f(a, b, c): 
    print(a, b, c)

f(1, 2, 3)          # 1 2 3
```

#### 2. Keywords

El orden de izquierda a derecha de los argumentos ya no importa cuando se utilizan palabras clave (keywords), porque los **argumentos se emparejan por nombre, no por posición**.


```python
def f(a, b, c): 
    print(a, b, c)

f(c=3, b=2, a=1)    # 1 2 3
```

```python
f(1, c=3, b=2)		# a gets 1 by position, b and c passed by name
                    # 1 2 3

# they make your calls a bit more self documenting
# f(name='Bob', age=40, job='dev')
```

#### 3. Defaults

Los valores por defecto nos permiten hacer que ciertos **argumentos de una función sean opcionales**; si no se les pasa un valor, el argumento recibe su valor predeterminado antes de que se ejecute la función.


```python

def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)


f(1)        # 1 2 3
f(a=1)      # 1 2 3
```

Si pasamos dos valores, solo `c` recibe su valor por defecto, y si pasamos tres valores, no se utiliza ningún valor por defecto.


```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)
    
f(1, 4)     # 1 4 3
f(1, 4, 5)	# Override defaults: 1 4 5

f(1, c=6)	# Choose defaults: 1 2 6
```

#### 4. Combining keywords and defaults
Observa nuevamente que cuando se usan argumentos con nombre (keyword arguments) en la llamada, el orden en que se enumeran los argumentos no importa; **Python los empareja por nombre, no por posición**.

El que llama a la función debe proporcionar valores para `spam` y `eggs`, pero estos pueden coincidir por posición o por nombre.

```python

def func(spam, eggs, toast=0, ham=0):
	print((spam, eggs, toast, ham)) 	# First 2 required

func(1, 2)							# Output: (1, 2, 0, 0)
func(1, ham=1, eggs=0)				# Output: (1, 0, 0, 1)
func(spam=1, eggs=0)				# Output: (1, 0, 0, 0)
func(toast=1, eggs=2, spam=3)		# Output: (3, 2, 1, 0)
func(1, 2, 3, 4) 					# Output: (1, 2, 3, 4)
```

### 5. Python 3.X Keyword-Only Arguments

También podemos usar un carácter `*` por sí solo en la lista de argumentos para indicar que una función no acepta una lista de argumentos de longitud variable, pero sí espera que todos los argumentos que siguen al `*` se pasen como argumentos con nombre (keywords).

```python
def kwonly(a, *b, c):
    print(a, b, c)

kwonly(1, 2, c=3)       # 1 (2,) 3

kwonly(a=1, c=3)        # 1 () 3

kwonly(1, 2, 3)         # TypeError: kwonly() missing 1 required keyword-only argument: 'c'
```

Todavía puedes usar valores por defecto para los argumentos que solo se pueden pasar por nombre (keyword-only), incluso si aparecen después del `*` en la cabecera de la función:

```python

def kwonly(a, *, b='spam', c='ham'):
	print(a, b, c)

kwonly(1)               # 1 spam ham
kwonly(1, c=3)          # 1 spam 3
kwonly(a=1)             # 1 spam ham
kwonly(c=3, b=2, a=1)   # 1 2 3
kwonly(1, 2)            # TypeError: kwonly() takes 1 positional argument but 2 were given
```

```python

def kwonly(a, *, b, c='spam'):
    print(a, b, c)

kwonly(1, c='eggs')     # TypeError: kwonly() missing 1 required keyword-only argument: 'b'
```

## Arbitrary Arguments Examples
Las dos últimas extensiones de emparejamiento, `*` y `**`, están diseñadas para admitir funciones que aceptan **cualquier cantidad de argumento**s.

Ambas pueden aparecer tanto en la **definición** de una función como en una **llamada** a función.

### Headers: Collecting arguments
En la _definición: de una función, el operador _recoge_ los argumentos posicionales no emparejados en una **tupla** y asigna esa tupla a la variable **args**.

```python
def f(*args):
    print(args)

f()                 # ()
f(1,)               # (1,)
f(1, 2, 3, 4)       # (1, 2, 3, 4)  
```

La característica `**` solo funciona con argumentos con nombre (keyword arguments). Recoge todos los argumentos que no coinciden explícitamente en la definición de la función y los agrupa en un nuevo **diccionario**.

```python
def f(**args):
    print(args)
f()                 # {}
f(a=1, b=2)         # {'a': 1, 'b': 2}
```

```python
def f(a, *pargs, **kargs):
    print(a, pargs, kargs)
f(1, 2, 3, x=1, y=2)        # 1 (2, 3) {'y': 2, 'x': 1}
```

### Calls: Unpacking arguments
La sintaxis `*` en una llamada a función se utiliza para desempaquetar una colección de argumentos posicionales.

```python
def func(a, b, c, d): print(a, b, c, d)

args = (1, 2)
args += (3, 4)
func(*args)         # Same as func(1, 2, 3, 4)
```

```python
args = {'a': 1, 'b': 2, 'c': 3}
args['d'] = 4
func(**args)    # Same as func(a=1, b=2, c=3, d=4)
```

```python
func(*(1, 2), **{'d': 4, 'c': 3})       # Same as func(1, 2, d=4, c=3)
# 1 2 3 4
func(1, *(2, 3), **{'d': 4})            # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, c=3, *(2,), **{'d': 4})         # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
func(1, *(2, 3), d=4)                   # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, *(2,), c=3, **{'d':4})      # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
```
