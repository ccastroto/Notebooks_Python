# EXCEPCIONES

### TRY / EXCEPT

```python
sep = '-' * 45 + '\n'
```

```python
print(sep + 'EXCEPTION RAISED AND CAUGHT')
try:
    x = 'spam'[99]
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

***EXCEPTION RAISED AND CAUGHT***

En el *try* intenta buscar la posición 99 en el string spam, saltará un **IndexError** porque no existe lo que se busca.

Entrará en el *except* por el 'IndexError' y saldrá lo que haya por haber saltado el **IndexError**, en este caso imprimirá el print.

Entrará el *finally* porque siempre que estea entrará y sacará lo que sea, en este caso imprimirá el print.

Tras las excepciones imprime el print final que se ejecuta al final de pasar por todo el try/except.

```python
print(sep + 'NO EXCEPTION RAISED')
try:
    x = 'spam'[3]
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

***NO EXCEPTION RAISED***

En el *try* buscará la posición 3 dentro del string spam, lo encontrará y no saltará nada.

No entra al *except* porque no saltó un *IndexError*, lo omite.

Entrará en el *finally* poruqe siempre que esté entra y sacará lo que sea.

Tras las excepciones imprime el print final que se ejecuta al final de pasar por todo el try/except.

```python
print(sep + 'NO EXCEPTION RAISED, WITH ELSE')
try:
    x = 'spam'[3]
except IndexError:
    print('exceprt run')
else:
    print('else run')
finally:
    print('finally run')
print('after run')
```

***NO EXCEPTION RAISED, WITH ELSE***

En el *try* buscará la posición 3 dentro del string spam, lo encontrará y no saltará nada.

No entra en el *except* poruqe no saltó un *IndexError*, lo omite.

Entra en el *else* porque no hubo una excepción (si hubiera no entraría en el else) y sacará lo que sea.

Entrará en el *finally* poruqe siempre que esté entra y sacará lo que sea.

Tras las excepciones imprime el print final que se ejecuta al final de pasar por todo el try/except.

```python
print(sep + 'EXCEPTION RAISED BUT NOT CAUGHT')
try:
    x = 1 / 0
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

***EXCEPTION RAISED BUT NOT CAUGHT***

Lo que hace en el *try* es intentar hacar una división por cero y sacará un **ZeroDivisionError**.

No entra en el *except* porque no está diseñada para el *ZeroDivisionError*, es decir, está diseñada para el *IndexError* y tras que no hay un IndexError no interactúa.

Entrará en el *finally* poruqe siempre que esté entra y sacará lo que sea.

Tras las excepciones no imprimirá el print final porque saltaría un error de *ZeroDivisionError* porque no está capturada por ninguna de las excepciones y saltará el error.
