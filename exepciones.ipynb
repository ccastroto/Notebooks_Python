# Unified try

```python
sep = '-' * 45 + '\n'
```

```python
print(sep + 'EXCEPTION RAISED AND CAUGHT')
```

```python
try:
    x = 'spam'[99]
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

```python
print(sep + 'NO EXCEPTION RAISED')
```

```python
try:
    x = 'spam'[3]
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

```python
print(sep + 'NO EXCEPTION RAISED, WITH ELSE')
```

```python
try:
    x = 'spam'[3]
except IndexError:
    print('except run')
else:
    print('else run')
finally:
    print('finally run')
print('after run')
```

```python
print(sep + 'EXCEPTION RAISED BUT NOT CAUGHT')
```

```python
try:
    x = 1 / 0
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```
