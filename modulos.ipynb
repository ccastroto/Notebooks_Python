### Sentencia import

La setencia import lista uno o más nombres de módulos para cargar separados por comas.

```python
import src.modulo as modulo
modulo.printer("Hello world")

# Hello world
```

### Sentencia from

from copia nombres específicos de un fichero a otro ámbito.

from permite listar uno o más nombres para ser copiados separados por comas.

```python
from src.modulo import printer
printer("Hello world")

# Hello world
```

### Sentencia from *

Cuando usamos * en cambio de nombres específicos obtenemos copias de todos los nombres asignados al ámbito global del módulo referenciado.

```python
from src.modulo import *
printer("Hello world")

# Hello world
```

### Cambiando mutables en modulos

También como def, the import and from son asignaciones implícitas:

import asigna un objeto módulo entero a un único nombre.
from asigna uno o más nombres a objetos con los mismos nombres en otro módulo.

```python
x = 1
y = [1, 2]
```

```python
from src.modulo_small import x, y
x = 42
y[0] = 42
x, y

# (42, [42, 2])
```

<b>x</b> no es un objeto compartido mutable, pero <b>y</b> sí:

```python
import src.modulo_small as small
small.x
# 1
small.x, small.y
# small.y == [42, 2]

# (1, [42, 2])
```

### Cuando import es requerido

La única vez que realmente debes usar import en lugar de from es cuando debes usar el mismo nombre definido en dos módulos diferentes.

```python
# M.py
def func():
...do something...
```

```python
# N.py
def func():
...do something else...
```

```python
# O.py
from M import func
from N import func
```

```python
# O.py
import M, N
M.func()
N.func()
```

## Mixed Usage Modes: name and main

Cada módulo tiene un atributo incorporado llamado nombre, que Python crea y asigna automáticamente de la siguiente manera:

-   Si el archivo se está ejecutando como un archivo de programa de nivel      
    superior, name se establece en la cadena «main» cuando se inicia.
-   Si, por el contrario, el archivo se está importando, name se asigna al     
    nombre del módulo tal y como lo conocen sus clientes.

El resultado es que un módulo puede comprobar su propio nombre para determinar si se está ejecutando o importando.

```python
def tester():
    print("It's Chritmas in Heaven...")

if __name == '__main__':
    tester()
```

Este módulo define una función para que los clientes la importen y utilicen como de costumbre:

```python
import runme
runme.tester()
# It's Chritmas in Heaven...
```

Pero el módulo también incluye código en la parte inferior que está configurado para llamar a la función automáticamente cuando este archivo se ejecuta como un programa:


```python
c:\code> python runme.py 
# It's Chritmas in Heaven...
```

En efecto, la variable de nombre de un módulo sirve como una bandera de modo de uso, permitiendo que su código sea aprovechado tanto como una biblioteca importable como un script de nivel superior.
